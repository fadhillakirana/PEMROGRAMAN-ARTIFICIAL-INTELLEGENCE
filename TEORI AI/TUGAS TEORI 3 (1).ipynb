```python
#menghitung rata-rata dari 3 bilangan bulat

a = int(input("Masukkan bilangan pertama: "))
b = int(input("Masukkan bilangan kedua: "))
c = int(input("Masukkan bilangan ketiga: "))

rata_rata = (a + b + c) / 3

print("Rata-rata =", rata_rata)
```

```python
#menghitung nilai fungsi f(x)

x = int(input("Masukkan nilai x: "))

if x != 0:
    fx = 2 * x**3 + 2 * x + 15 / x
    print("Nilai f(x) =", fx)
else:
    print("Nilai x tidak boleh 0")
```

```python
#gambar 3.3

umur = int(input("Umur Anda: "))

if umur <= 5:
    kriteria = "Balita"
elif umur <= 13:
    kriteria = "Anak-anak"
elif umur <= 25:
    kriteria = "Remaja"
elif umur <= 35:
    kriteria = "Dewasa"
elif umur <= 55:
    kriteria = "Orang Tua"
else:
    kriteria = "Lansia"

print("Umur %d dan kriteria yaitu %s" % (umur, kriteria))
```

```python

# PERINTAH IF BERSARANG

Nama = input('Nama Karyawan: ')
NIK = input('NIK Karyawan: ')
Status = input('Status (Staff/Honor): ')
Gol = input('Golongan (A/B/C): ')

if Status == 'Staff':
    Gaji = 1000000

    if Gol == 'A':
        Bonus = 200000
    elif Gol == 'B':
        Bonus = 400000
    elif Gol == 'C':
        Bonus = 550000
    else:
        Bonus = 0

elif Status == 'Honor':
    Gaji = 750000

    if Gol == 'A':
        Bonus = 150000
    elif Gol == 'B':
        Bonus = 275000
    elif Gol == 'C':
        Bonus = 480000
    else:
        Bonus = 0

else:
    Gaji = 0
    Bonus = 0
    print('Status karyawan tidak valid')

Gatot = Gaji + Bonus

print('Nama Karyawan:', Nama)
print('NIK Karyawan:', NIK)
print('Gaji Rp', Gaji)
print('Bonus Rp', Bonus)
print('Gaji Total Rp', Gatot)
```
