```python
#Contoh pengulangan for sederhana
angka = [1,2,3,4,5]
for x in angka:
    print(x)

#Contoh pengulangan for
nama = ["FADHILLA", "KIRANA", "UTAMI"]
for panggilan in nama:
    print ("Anda bisa memanggil saya", panggilan)
```

```python
#Contoh penggunaan Nested Loop

i = 5
while(i < 100):
    j = 3
    while(j <= (i/j)):
        if not(i%j): break
        j = j + 1
    if (j > i/j) : print(i, " ARCHON QUEST 7.2 IS PEAK")
    i = i + 1

print("SEE U IN ARCHON QUEST 7.3")
```
